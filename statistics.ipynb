*   U критерий Вилкоксона-Манна-Уитни & rank-biserial correlation
*   критерий χ² Пирсона & V Крамера



```python
from scipy.stats import ttest_ind, mannwhitneyu
import numpy as np

col = 'название колонки df, соответствующей параметру'

#t-критерий Стьюдента
x1 = df[df['GROUP'] == 1.0][col].dropna()
x2 = df[df['GROUP'] == 2.0][col].dropna()
stat, p =  ttest_ind(x1, x2)
print(stat, p)

#d Коэна
def cohens_d(x1, x2):
    n1 = len(x1)
    n2 = len(x2)
    s1 = np.var(x1, ddof=1)
    s2 = np.var(x2, ddof=1)
    pooled_sd = np.sqrt(((n1 - 1)*s1 + (n2 - 1)*s2) / (n1 + n2 - 2))
    d = (np.mean(x1) - np.mean(x2)) / pooled_sd

    return d

d = cohens_d(x1, x2)
print(d)
```

```python
col = 'название колонки df, соответствующей параметру'

#U критерий Вилкоксона-Манна-Уитни
x1 = df[df['GROUP'] == 1.0][col].dropna()
x2 = df[df['GROUP'] == 2.0][col].dropna()
stat, p =  mannwhitneyu(x1, x2)

#ank-biserial correlation
n1 = len(x1)
n2 = len(x2)
rbc = 1 - (2 * stat) / (n1 * n2)

print(stat, p)
print(rbc)
```
